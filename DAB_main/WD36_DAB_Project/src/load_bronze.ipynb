```python
dbutils.widgets.text("catalog", "defaultcatalog")
dbutils.widgets.text("schema", "defaultschema")
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
```

```python
data = [(1, "a"), (2, "b"), (3, "c"), (4, "d")]
spark.createDataFrame(data).toDF("id", "value").write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.sampletable")
```
